# 一般参加者の臭気快価性に関連する知覚構造の分析
## 参加者外予測・主成分分析・アブレーション分析による検討


# 要旨

本分析の目的は、一般参加者が臭気に対して示す快・不快度が、食用らしさ、親しみ、温冷感、刺激性、嫌悪感、質的な臭気記述などとどのような関係を持つかを明らかにすることである。また、その関係が特定の参加者だけに依存するのではなく、学習に使用していない参加者にも一定程度共通して確認できる構造であるかを検討した。

Bierlingらが公開した、1,227名による74種類の単一分子臭気評価データを使用した[1]。快・不快度`pleasant`の連続値を線形回帰で推定し、`pleasant>=50`を快、`pleasant<50`を非快としてロジスティック回帰で分類した。また、8種類の連続的な知覚評価にPCAを適用した。参加者固有の回答傾向が訓練データからテストデータへ混入することを防ぐため、被験者IDを単位として訓練データとテストデータを分割した。

線形回帰はMAE 14.15、RMSE 18.69、決定係数$$R^2=0.614$$となり、平均値ベースラインよりMAEを45.3%、RMSEを37.9%低減した。ロジスティック回帰はAccuracy 82.6%、F1 82.8%、ROC-AUC 0.905となった。PCAでは、`pleasant`、`edible`、`warm`、`familiar`が同じ方向に、`disgusting`、`irritating`、`cold`が反対方向に配置される第一主成分が得られた。

さらに、入力情報を段階的に追加するアブレーション分析を行った。年齢・性別・臭気物質コードだけでは$$R^2=0.174$$、ROC-AUC 0.697であったが、16種類の質的ラベルを追加すると$$R^2=0.368$$、ROC-AUC 0.815となった。`disgusting`を除く連続知覚評価まで追加すると$$R^2=0.520$$、ROC-AUC 0.873となり、`disgusting`を追加した完全モデルでは$$R^2=0.614$$、ROC-AUC 0.905となった。一方、完全モデルから臭気物質コードを除外しても$$R^2=0.602$$、ROC-AUC 0.905であり、性能の大部分は特定の臭気物質コードの記憶ではなく、回答時に形成された知覚内容から得られていた。

以上から、本データにおける臭気の快・不快度は完全に独立した一項目ではなく、意味的な臭気表現、食用らしさ、親しみ、温かさ、嫌悪感、刺激性などが段階的に重なった知覚表現の一部であることが示唆された。ただし、本分析は同時に回答された主観評価から`pleasant`を再構成したものであり、分子構造やセンサ値から未知臭気の快度を事前予測したものではない。

# 1．背景

## 1.1 臭気知覚の構造と個人差

臭気物質の分子構造は、人間がその臭気をどのように知覚するかに関係する。しかし、化学構造から「甘い」「果物らしい」「快い」「刺激的」といった知覚を一意に説明する一般則は十分には確立されていない。さらに、臭気知覚は年齢、性別、文化的背景、経験、記憶、測定状況などの影響を受けるため、同じ臭気物質であっても評価には個人差が生じる。

従来の臭気データベースには、調香師などの専門家による評価を中心としたものが多い。専門家の評価は再現性が高い一方、一般の人々の日常的な臭気知覚をそのまま代表するとは限らない。また、評価前に用意された記述語を提示すると、その語が参加者の注意や判断を誘導する可能性がある。

Bierlingらは、この問題に対して、一般参加者による大規模な臭気知覚データセットを構築した[1]。参加者は、定型的な評価項目を回答する前に臭気を自由記述し、その後に8種類の連続評価と16種類の質的ラベルを回答した。この順序により、定型ラベルによる知覚の誘導を抑えながら、自由な意味表現と標準化された尺度の両方を取得している。

## 1.2 原著の実験計画

原著では、1人が74種類すべてを評価する完全交差型の実験ではなく、参加者を9群に分ける入れ子型の計画が採用された。各参加者は10種類の臭気を評価し、その内訳は群ごとに異なる8種類の臭気と、すべての群に共通するbenzyl acetateおよび4-decanolideの2種類のアンカー臭気であった[1]。

この設計は、多数の臭気を連続して嗅ぐことによる嗅覚順応や臭気間干渉を抑えながら、大規模データを収集するための合理的な方法である。一方、各臭気を評価した参加者構成は完全には同一ではない。そのため、臭気物質別平均の差には物質差だけでなく、群構成や測定環境の影響が含まれる可能性がある。

原著では、臭気物質間の強度差を抑えるため、アンカー臭気を基準として各臭気の濃度を事前調整している[1]。また、別の一般参加者データとの比較、アンカー臭気による自宅群と実験室群の比較、60名による約1週間後の再検査を用いて、妥当性、環境への頑健性、時間的安定性を検証している。再検査ではpleasantnessが$$r=0.81$$、edibilityが$$r=0.79$$、disgustが$$r=0.77$$であり、快価性に関係する評価は比較的高い時間的安定性を示した[1]。

## 1.3 研究課題

本分析では、機械学習手法の性能比較そのものではなく、次の研究課題を設定した。

1. 臭気の快・不快度は、他の知覚評価および質的ラベルとどのように関連するか。
2. その関係は、学習に使用していない参加者にも適用できる程度に共有されているか。
3. 高い予測性能は、臭気物質の種類、質的ラベル、連続知覚評価、嫌悪感のどの情報から生じるか。
4. 快・非快という結論は、二値化閾値を40、50、60へ変えても維持されるか。
5. 複数の知覚評価の背後に、快価性や知覚的顕著性に対応する潜在軸が存在するか。

これらの問いに答えるため、線形回帰、ロジスティック回帰、PCA、アブレーション分析、閾値感度分析を用いた。

# 2．使用データ

## 2.1 臭気評価項目

参加者は臭気を自由記述した後、次の8項目を視覚的アナログ尺度で評価した。

- `pleasant`：快・不快度
- `intensive`：臭気の強度
- `familiar`：臭気への親しみ
- `warm`：温かい印象
- `cold`：冷たい印象
- `irritating`：刺激性
- `edible`：食用らしさ
- `disgusting`：嫌悪感

さらに、`sweet`、`sour`、`fruit`、`spices`、`bakery`、`garlic`、`fish`、`burnt`、`decayed`、`grass`、`wood`、`chemical`、`flower`、`musky`、`sweaty`、`ammonia/urinous`の16項目を二値で回答した。

## 2.2 使用ファイルと分析対象

本分析では、Zenodoで公開されたVersion 3の次の3ファイルを使用した[2]。

- `data.csv`
- `free_descriptions_translated.csv`
- `odors.csv`

`data.csv`はロング形式であり、1行が1人の参加者による1種類の臭気評価を表す。全データは13,260行、1,314名分であった。原著の採用基準を満たす`inclusion==1`のデータは12,390行、1,227名分であった。`pleasant`と被験者IDが欠損していない機械学習用データは12,005行、1,202名分であった。

`free_descriptions_translated.csv`には、出現頻度上位100位以内のドイツ語記述語と英訳が収録されている。これは原自由記述37,041件のうち26,115件、70.5%をカバーしたものであり、全自由記述を完全に翻訳したデータではない[1]。

なお、原著では74種類の臭気刺激が選定されたが、tributyrinとcuminolはプラスチック容器との反応を理由に2-phenylethanolへ置換された。本分析で使用した評価データ上では73種類の`molcode`が確認された。

# 3．分析方法

## 3.1 分析の位置付け

本分析で予測する`pleasant`は、臭気を嗅いだ後に得られた主観評価である。また、`disgusting`、`edible`、`warm`などの説明変数も同じ評価場面で得られている。

したがって、本分析は、分子構造やセンサ値から未知臭気の快度を事前に予測する分析ではない。複数の同時的な知覚評価から`pleasant`を再構成することにより、臭気知覚の内部構造を数量化する分析である。

これは訓練データとテストデータの混入という技術的なデータリーケージではないが、目的変数と説明変数の間には構成概念上の近さがある。特に`disgusting`は`pleasant`と反対方向の意味を持つため、完全モデルの性能だけでは、何の情報によって予測が可能になったかを判断できない。そこでアブレーション分析を追加した。

## 3.2 前処理と参加者単位分割

連続特徴量には`intensive`、`familiar`、`warm`、`cold`、`irritating`、`edible`、`disgusting`、`age`を使用した。16種類の質的ラベルは0または1の二値特徴量として使用し、`molcode`と`sex`はカテゴリ変数としてOne-Hot Encodingを適用した。

数値特徴量の欠損値は訓練データの中央値で補完し、平均0、標準偏差1に標準化した。前処理は`Pipeline`内で実行し、テストデータの情報が学習に使用されることを防いだ。

同一参加者は複数臭気を評価しているため、被験者ID`code`をグループとして`GroupShuffleSplit`を用い、訓練80%、テスト20%に分割した。訓練データは9,584行、961名、テストデータは2,421行、241名となり、被験者の重複は0名であった。

この分割が評価するのは未知の参加者への汎化である。臭気物質は訓練側とテスト側に共通して存在するため、未知分子への外挿性能を評価するものではない。

## 3.3 線形回帰

線形回帰では`pleasant`を連続値として推定した。

$$\hat{y}=\beta_0+\beta_1x_1+\beta_2x_2+\cdots+\beta_px_p$$

評価にはMAE、RMSE、決定係数$$R^2$$を使用し、訓練データの平均値を常に出力する平均値ベースラインと比較した。

## 3.4 ロジスティック回帰

`pleasant>=50`を1（快）、`pleasant<50`を0（非快）として二値分類した。

$$P(y=1\mid x)=\frac{1}{1+\exp\left[-\left(\beta_0+\beta_1x_1+\cdots+\beta_px_p\right)\right]}$$

評価にはAccuracy、Precision、Recall、F1、ROC-AUC、Balanced Accuracy、混同行列を用いた。また、訓練データの多数派クラスのみを出力する多数派ベースラインと比較した。

## 3.5 主成分分析

PCAでは、`pleasant`、`intensive`、`familiar`、`warm`、`cold`、`irritating`、`edible`、`disgusting`の8項目を標準化して分析した。寄与率により情報の圧縮度を確認し、負荷量から各主成分の知覚的意味を解釈した。

## 3.6 アブレーション分析

予測性能がどの情報から生じたかを検討するため、同一の訓練・テスト分割で次の5モデルを比較した。

- A：年齢、性別、臭気物質コード
- B：A＋16種類の質的ラベル
- C：B＋`disgusting`を除く連続知覚評価
- D：C＋`disgusting`（完全モデル）
- E：完全モデルから臭気物質コードを除外

AからDへの変化により、情報を追加したときの性能向上を検討した。DとEの比較により、完全モデルが臭気物質コードを記憶することへどの程度依存しているかを確認した。

## 3.7 二値化閾値の感度分析

50点という境界は尺度の中点ではあるが、確立された生理学的境界ではない。そこで、快の定義を`pleasant>=40`、`pleasant>=50`、`pleasant>=60`と変え、Accuracy、Balanced Accuracy、F1、ROC-AUCを比較した。

# 4．結果

## 4.1 基本統計量と単変量関係

`pleasant`の平均値は47.61、標準偏差は29.82、中央値は49であり、1から100まで広く分布した。50点以上を快と定義した場合、快の割合は約49.8%であった。

Spearman順位相関では、`pleasant`と`disgusting`の相関が-0.664で最も強く、`edible`は0.495、`warm`は0.394、`familiar`は0.368、`irritating`は-0.356であった。したがって、快度は嫌悪感と強く反対方向に変化し、食用らしさ、温かさ、親しみとは同方向に変化する傾向が確認された。ただし、この相関は1人から複数行を得た階層構造を考慮しない記述統計である。

## 4.2 線形回帰

平均値ベースラインはMAE 25.88、RMSE 30.10、$$R^2$$はほぼ0であった。完全な線形回帰モデルはMAE 14.15、RMSE 18.69、$$R^2=0.614$$となり、ベースラインよりMAEを45.3%、RMSEを37.9%低減した。

図2では実測値と予測値に正の対応が確認された。一方、極端に高い評価と低い評価では、予測値が中央へ近づく傾向も見られた。図3の残差は概ね0を中心に分布したが、RMSEは18.69であり、共有された規則性だけでは説明できない個人差や未観測要因も大きい。

標準化した数値・二値特徴量の係数では、`edible`、`warm`、`familiar`が`pleasant`を高くする方向に関連し、`disgusting`は係数-12.34となって最も強い負方向の関連を示した。ただし、係数は他の変数を一定とした条件付きの関係であり、因果効果ではない。

## 4.3 ロジスティック回帰

多数派ベースラインのAccuracyは48.6%であった。完全モデルはAccuracy 82.6%、Precision 79.8%、Recall 85.9%、F1 82.8%、ROC-AUC 0.905となった。

混同行列では、非快を正しく非快と分類した件数が990件、快を正しく快と分類した件数が1,010件であった。非快を快と誤分類した件数は255件、快を非快と誤分類した件数は166件であった。快に対する感度は85.9%、非快に対する特異度は79.5%、Balanced Accuracyは82.7%であり、一方のクラスだけに偏った性能ではなかった。

## 4.4 主成分分析

第一主成分の寄与率は36.1%、第二主成分は20.3%であり、累積寄与率は56.4%であった。

第一主成分では、`pleasant`が0.484、`edible`が0.394、`warm`が0.376、`familiar`が0.241であり、`disgusting`が-0.450、`irritating`が-0.345、`cold`が-0.272であった。このことから、第一主成分は「食用らしい・親しみがある・温かい・快い」という側面と、「嫌悪感がある・刺激的・冷たい・不快である」という側面を対立させる快価性軸と解釈した。

第二主成分では、`intensive`が0.641、`familiar`が0.565、`irritating`が0.402であった。この軸は、臭気がどの程度明瞭に知覚され、認識され、刺激として目立つかを表す知覚的顕著性の軸と解釈した。

第一・第二主成分で説明できない分散は43.6%残っているため、臭気知覚を快不快と強度の二軸だけへ還元することはできない。

## 4.5 アブレーション分析

年齢・性別・臭気物質コードのみを用いたモデルAは、線形回帰で$$R^2=0.174$$、ロジスティック回帰でROC-AUC 0.697であった。臭気物質の種類と属性だけでも平均的な快度傾向をある程度捉えられるが、十分ではなかった。

16種類の質的ラベルを追加したモデルBでは、$$R^2=0.368$$、ROC-AUC 0.815となった。したがって、「甘い」「果物らしい」「化学的」などの意味的ラベルは、臭気物質コードだけでは表現できない知覚情報を含んでいた。

`disgusting`を除く連続知覚評価を追加したモデルCでは、$$R^2=0.520$$、ROC-AUC 0.873となった。親しみ、温冷感、刺激性、食用らしさなどが、快度の連続的な差を説明する重要な情報であることが分かった。

`disgusting`を追加した完全モデルDでは、$$R^2=0.614$$、ROC-AUC 0.905となった。モデルCからDへの増加は、$$R^2$$で0.094、ROC-AUCで0.032であった。嫌悪感は重要であるが、嫌悪感を除外した時点でも相当の性能が残っていた。したがって、快度予測が単に`pleasant`の反対語である`disgusting`だけに依存しているわけではない。

完全モデルから臭気物質コードを除外したモデルEでも、$$R^2=0.602$$、Accuracy 82.2%、ROC-AUC 0.905であった。完全モデルとの差は小さく、高い性能の大部分は特定の臭気物質コードの記憶ではなく、参加者がその臭気をどのように知覚したかという回答内容から生じていた。

以上から、快度を説明する情報は、臭気物質の平均的傾向、意味的な質的ラベル、連続的な知覚評価、嫌悪感という順に階層的に追加されると考えられる。

## 4.6 二値化閾値の感度分析

快の閾値を40、50、60へ変更した場合、ROC-AUCはそれぞれ0.897、0.905、0.902、Balanced Accuracyは0.803、0.827、0.828であった。

したがって、ロジスティック回帰が快と非快を識別できるという結論は、50点という単一の閾値だけに依存していなかった。一方、Precision、Recall、クラス比率は閾値によって変化するため、実用的な分類器では目的に応じた閾値設定が必要である。

## 4.7 臭気物質別の平均快度

平均`pleasant`が高かった臭気物質は、`(R)-(-)-carvone`が76.95、`maltol`が74.94、`eucalyptol`が72.25であった。低かった臭気物質は、`p-tolyl acetate`が17.31、`2-isopropyl-5-methylanisole`が22.10、`caproic acid`が24.82であった。

最上位と最下位には約59.6点の差があり、臭気物質の種類と平均快度には大きな関係があった。原著では再検査刺激の選定時に、`maltol`が最も快く刺激性が低い臭気、`(R)-(-)-carvone`が最も親しみが高く嫌悪感が低い臭気、`p-tolyl acetate`が最も不快で刺激性と嫌悪感が高い臭気として選ばれており、本分析の集計と整合した[1]。

ただし、入れ子型設計により各物質を評価した参加者は同一ではないため、この順位を物質固有の普遍的性質と断定することはできない。

## 4.8 自由記述

翻訳済み自由記述では、`sweet`が2,634件、`NEG-pleasant`が1,304件、`fruity`が1,068件、`chemical`が991件、`pleasant`が979件であった。

食品・自然物に関連する`sweet`、`fruity`、`flowery`、`fresh`と、人工性・刺激性に関連する`chemical`、`artificial`、`stinging`がともに高頻度で現れた。これは、臭気知覚が単純な快・不快だけではなく、食品との関連、自然らしさ、人工性、刺激性、強度などを含む多次元的な意味表現であることを示している。

# 5．考察

## 5.1 共有構造と個人差の共存

参加者単位で訓練・テストを分けても、線形回帰とロジスティック回帰はベースラインを大きく上回った。したがって、臭気の快・不快判断は完全に個人的で無秩序なものではなく、他の知覚評価との間に参加者を越えて共有される統計的構造が存在すると考えられる。

一方、線形回帰のRMSEは18.69であり、完全な再構成には至っていない。PCAでも第一・第二主成分以外に43.6%の分散が残った。したがって、共有構造が存在する一方で、経験、文化、記憶、体調、測定状況などによる個人差と多次元性も大きい。

## 5.2 快度は単一尺度ではなく階層的な知覚表現である

アブレーション分析により、予測性能は臭気物質コードだけで突然得られたものではなく、質的ラベル、連続知覚評価、嫌悪感を追加するごとに段階的に向上した。

この結果は、快・不快度が一つの孤立した回答ではなく、まず「何のような臭気か」という意味的分類があり、その上に親しみ、温冷感、刺激性、食用らしさなどの連続的評価が形成され、最終的な快価性と強く結び付くという階層的な表現を持つ可能性を示す。

ただし、データは同一時点の回答であり、この順序を時間的・因果的な処理順序として証明したものではない。ここで示したのは、情報量が段階的に追加される統計的構造である。

## 5.3 嫌悪感は中心的だが唯一の説明ではない

`disgusting`は最も強い負の相関と回帰係数を示し、完全モデルの性能を大きく改善した。したがって、嫌悪感は臭気快度における中心的な知覚要素である。

しかし、`disgusting`を除いたモデルCでも$$R^2=0.520$$、ROC-AUC 0.873であった。食用らしさ、親しみ、温冷感、刺激性、質的ラベルだけでも快度の相当部分を再構成できた。したがって、快度は単に嫌悪感の反転としてだけではなく、複数の知覚的・意味的特徴の組合せとして表現されている。

## 5.4 臭気物質コードへの依存は限定的である

完全モデルから`molcode`を除外しても性能はほとんど低下しなかった。この結果は、完全モデルが「この物質は平均的に快い」という物質IDの対応表を記憶しただけではないことを示す。

同じ臭気物質でも参加者ごとに評価が異なる中で、モデルは各参加者が回答した意味的ラベルや知覚評価を利用して快度を再構成していたと考えられる。これは、臭気物質固有の平均傾向と、参加者ごとの知覚状態を区別するうえで重要である。

## 5.5 原著の妥当性検証との整合

原著では、別の一般参加者データとの比較においてpleasantness、familiarity、edibility、intensityに有意な正の関係が確認された。また、再検査ではpleasantness、edibility、disgustが比較的高い時間的安定性を示した[1]。

本分析で`pleasant`、`edible`、`disgusting`が同じ快価性軸上に現れ、未知参加者にも一般化したことは、原著の妥当性・再検査結果と整合する。したがって、得られた快価性構造は単なる一時的な回答ノイズだけで形成されたとは考えにくい。

一方、原著では自宅群と実験室群の間でintensityとcoldnessに小さいながら有意差があった[1]。本分析の第二主成分に現れた強度・冷たさの一部には、物質差だけでなく測定環境差が反映されている可能性がある。

## 5.6 本分析から言えることと言えないこと

本分析から言えるのは、本データにおける一般参加者の臭気快度が、意味的ラベル、食用らしさ、親しみ、温冷感、刺激性、嫌悪感と結び付いた共有構造を持つことである。また、その構造は特定の臭気物質コードだけに依存せず、二値化閾値を変えても概ね維持された。

一方、次のことは本分析からは言えない。

- `edible`や`familiar`を高めれば因果的に`pleasant`が上昇すること
- 分子構造から未知臭気の快度を予測できること
- すべての文化圏・年齢層で同じ構造が成立すること
- 臭気評価の心理過程が、アブレーション分析のAからDの順に時間的に進行すること
- 個々の参加者の好みを高精度に完全予測できること

機械学習はこれらの結論を自動的に保証するものではなく、研究課題に対してデータ中の規則性を数量化する手段である。

# 6．限界と今後の課題

第一に、1人が複数臭気を評価し、1臭気を複数人が評価する階層構造を通常の線形回帰では完全には扱えていない。さらに、再検査参加者は二時点の回答を持つため、他の参加者より多くの行を占める。被験者と臭気物質をランダム効果とする混合効果モデルによって、個人差、物質差、測定時点差を分離する必要がある。

第二に、説明変数間の相関が強い。特に`pleasant`、`disgusting`、`edible`は概念的に近く、個々の係数は多重共線性の影響を受ける可能性がある。VIF、Ridge回帰、Lasso回帰、ブートストラップによる係数安定性の検証が必要である。

第三に、臭気物質コードは訓練・テスト間で共有されている。未知分子の予測には、臭気物質単位で分割し、SMILES、分子量、官能基、分子フィンガープリントなどの客観的特徴量を使用する必要がある。

第四に、原著の参加者は平均年齢26.18歳で女性が68.27%を占め、ドイツ語環境で評価されている[1]。文化、年齢、言語、臨床状態の異なる集団への一般化には外部検証が必要である。

第五に、自由記述は翻訳済み上位語の頻度分析に限定される。臭気物質別TF-IDF、文章埋め込み、クラスタリングを用いれば、数値尺度では捉えにくい意味構造を分析できる。

# 7．結論

本分析から、本データにおける一般参加者の臭気快度は、完全に個人的で無秩序な評価ではなく、臭気物質の平均的傾向、質的な意味ラベル、親しみ、温冷感、刺激性、食用らしさ、嫌悪感が段階的に重なった知覚構造を持つことが示唆された。

年齢・性別・臭気物質コードだけでも一定の予測力は存在したが、質的ラベルと連続知覚評価を追加することで性能は段階的に向上した。`disgusting`は中心的な情報であったが、それを除いても高い予測力が残った。また、臭気物質コードを除いても完全モデルの性能はほぼ維持された。

したがって、快・不快度は単なる物質IDの属性でも、嫌悪感の反転だけでもなく、参加者が臭気をどのような対象として意味付けし、どの程度親しみ、食用らしさ、温冷感、刺激性を感じたかを統合した知覚表現の一部であると考えられる。

一方、予測誤差とPCAの未説明分散は小さくなく、共有構造だけでは説明できない個人差も残る。また、本分析は臭気を嗅いだ後の同時回答を再構成したものであり、未知臭気の事前予測や因果関係を示したものではない。

以上より、本分析の主要な成果は、機械学習モデルの精度そのものではなく、臭気快価性を支える情報がどの層に存在し、それぞれがどの程度の情報を追加するかを数量的に示した点にある。

# 以下：分析コード

以下のセルを上から順番に実行することで、本文で参照した図および評価値を再現できる。


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from sklearn.compose import ColumnTransformer
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix,
    f1_score,
    mean_absolute_error,
    mean_squared_error,
    precision_score,
    r2_score,
    recall_score,
    roc_auc_score,
    roc_curve,
)
from sklearn.model_selection import GroupShuffleSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

plt.rcParams["font.family"] = "DejaVu Sans"

BASE_URL = "https://zenodo.org/records/15657278/files"
df = pd.read_csv(f"{BASE_URL}/data.csv?download=1", sep=";", encoding="utf-8-sig", low_memory=False)
df_f = pd.read_csv(
    f"{BASE_URL}/free_descriptions_translated.csv?download=1",
    sep=";",
    encoding="utf-8-sig",
    low_memory=False,
)
df_o = pd.read_csv(f"{BASE_URL}/odors.csv?download=1", sep=";", encoding="utf-8-sig", low_memory=False)

print("data.csv:", df.shape)
print("free_descriptions_translated.csv:", df_f.shape)
print("odors.csv:", df_o.shape)
```

# 1．データの基本確認

データセット作成者が分析対象として指定した`inclusion == 1`の回答を抽出する。

全データと分析対象データについて、行数、被験者数、臭気物質数、要約統計量および欠損数を確認し、`pleasant`の分布を図1に示す。


```python
RATING_COLUMNS = [
    "pleasant", "intensive", "familiar", "warm",
    "cold", "irritating", "edible", "disgusting",
]

BINARY_ODOR_COLUMNS = [
    "sweet", "sour", "fruit", "spices", "bakery", "garlic",
    "fish", "burnt", "decayed", "grass", "wood", "chemical",
    "flower", "musky", "sweaty", "ammonia/urinous",
]

numeric_columns = RATING_COLUMNS + BINARY_ODOR_COLUMNS + ["age", "sex", "inclusion"]
df[numeric_columns] = df[numeric_columns].apply(pd.to_numeric, errors="coerce")

included = df.loc[df["inclusion"].eq(1)].copy()

print("全データ:", len(df), "行 /", df["code"].nunique(), "名")
print("分析対象:", len(included), "行 /", included["code"].nunique(), "名")
print("臭気物質:", included["molcode"].nunique(), "種類")

display(included[RATING_COLUMNS].describe().T)
display(included[RATING_COLUMNS].isna().sum().rename("missing").to_frame())

pleasant_correlations = (
    included[RATING_COLUMNS]
    .corr(method="spearman")["pleasant"]
    .drop("pleasant")
    .sort_values(ascending=False)
    .rename("Spearman_rho")
    .to_frame()
)
display(pleasant_correlations)

plt.figure(figsize=(8, 5))
plt.hist(included["pleasant"].dropna(), bins=20, edgecolor="black")
plt.title("Figure 1: Distribution of pleasant ratings")
plt.xlabel("pleasant score")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

# 2．機械学習用データの準備

線形回帰では`pleasant`を連続値として予測し、ロジスティック回帰では`pleasant >= 50`を1（快）、それ未満を0（非快）とする。

説明変数には、臭気の知覚評価、16種類の質的ラベル、年齢、臭気物質コードおよび性別を使用する。`sex`と`molcode`は数値の大小に意味がないため、カテゴリ変数として扱う。


```python
CONTINUOUS_FEATURES = [
    "intensive", "familiar", "warm", "cold",
    "irritating", "edible", "disgusting", "age",
]
CATEGORICAL_FEATURES = ["molcode", "sex"]
MODEL_FEATURES = CONTINUOUS_FEATURES + BINARY_ODOR_COLUMNS + CATEGORICAL_FEATURES

model_data = included[["code", "pleasant", *MODEL_FEATURES]].dropna(
    subset=["code", "pleasant"]
).copy()

model_data["code"] = model_data["code"].astype(str).str.strip()
model_data["molcode"] = model_data["molcode"].astype(str).str.strip()
model_data["sex"] = pd.to_numeric(model_data["sex"], errors="coerce").map(
    {0: "0", 1: "1", 2: "2"}
)

X = model_data[MODEL_FEATURES]
y = model_data["pleasant"].astype(float)
groups = model_data["code"]

print("モデル用データ:", len(model_data), "行 /", groups.nunique(), "名")
print("説明変数:", len(MODEL_FEATURES), "項目")
```

## 被験者単位での訓練・テスト分割

同一被験者の回答が訓練データとテストデータの両方に含まれると、評価性能を過大に見積もる可能性がある。

そこで、`GroupShuffleSplit`を用いて被験者ID単位で訓練80%、テスト20%に分割する。前処理は`Pipeline`内で学習し、テストデータからの情報漏洩を防ぐ。


```python
RANDOM_STATE = 42

splitter = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=RANDOM_STATE)
train_index, test_index = next(splitter.split(X, y, groups=groups))

X_train, X_test = X.iloc[train_index], X.iloc[test_index]
y_train, y_test = y.iloc[train_index], y.iloc[test_index]
train_groups, test_groups = groups.iloc[train_index], groups.iloc[test_index]

print("訓練:", len(X_train), "行 /", train_groups.nunique(), "名")
print("テスト:", len(X_test), "行 /", test_groups.nunique(), "名")
print("被験者の重複:", len(set(train_groups) & set(test_groups)))

def make_preprocessor():
    numeric_features = CONTINUOUS_FEATURES + BINARY_ODOR_COLUMNS

    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ])
    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ])

    return ColumnTransformer([
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, CATEGORICAL_FEATURES),
    ])
```

# 3．線形回帰

`pleasant`を連続値として予測し、訓練データの平均値を常に予測するベースラインと比較する。

評価指標にはMAE、RMSE、$$R^2$$を使用する。数値結果に加え、実測値と予測値の散布図および残差分布を図2、図3に示す。


```python
linear_model = Pipeline([
    ("preprocessor", make_preprocessor()),
    ("model", LinearRegression()),
])
linear_model.fit(X_train, y_train)

linear_prediction = linear_model.predict(X_test)
baseline_prediction = np.full(len(y_test), y_train.mean())

linear_metrics = pd.DataFrame({
    "Model": ["Mean baseline", "Linear regression"],
    "MAE": [
        mean_absolute_error(y_test, baseline_prediction),
        mean_absolute_error(y_test, linear_prediction),
    ],
    "RMSE": [
        np.sqrt(mean_squared_error(y_test, baseline_prediction)),
        np.sqrt(mean_squared_error(y_test, linear_prediction)),
    ],
    "R2": [
        r2_score(y_test, baseline_prediction),
        r2_score(y_test, linear_prediction),
    ],
})
display(linear_metrics)
```

```python
linear_residual = y_test.to_numpy() - linear_prediction

plt.figure(figsize=(6, 6))
plt.scatter(y_test, linear_prediction, alpha=0.25, s=18)
plot_min = min(y_test.min(), linear_prediction.min())
plot_max = max(y_test.max(), linear_prediction.max())
plt.plot([plot_min, plot_max], [plot_min, plot_max], linestyle="--")
plt.title("Figure 2: Linear regression - actual vs predicted")
plt.xlabel("Actual pleasant")
plt.ylabel("Predicted pleasant")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.hist(linear_residual, bins=30, edgecolor="black")
plt.axvline(0, linestyle="--")
plt.title("Figure 3: Linear regression - residual distribution")
plt.xlabel("Actual - predicted")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
linear_coefficients = pd.DataFrame({
    "feature": linear_model.named_steps["preprocessor"].get_feature_names_out(),
    "coefficient": linear_model.named_steps["model"].coef_,
}).sort_values("coefficient", ascending=False)

linear_numeric_coefficients = linear_coefficients[
    linear_coefficients["feature"].str.startswith("numeric__")
].copy()

print("pleasantを高くする方向（数値・二値特徴量）")
display(linear_numeric_coefficients.head(10))

print("pleasantを低くする方向（数値・二値特徴量）")
display(linear_numeric_coefficients.tail(10).sort_values("coefficient"))
```

# 4．ロジスティック回帰

`pleasant >= 50`を1（快）、`pleasant < 50`を0（非快）として二値分類を行う。

Accuracy、Precision、Recall、F1、ROC-AUCを算出し、混同行列とROC曲線を図4、図5に示す。


```python
y_train_class = (y_train >= 50).astype(int)
y_test_class = (y_test >= 50).astype(int)

logistic_model = Pipeline([
    ("preprocessor", make_preprocessor()),
    ("model", LogisticRegression(max_iter=3000, random_state=RANDOM_STATE)),
])
logistic_model.fit(X_train, y_train_class)

logistic_probability = logistic_model.predict_proba(X_test)[:, 1]
logistic_prediction = (logistic_probability >= 0.5).astype(int)

majority_class = int(y_train_class.mode().iloc[0])
baseline_class_prediction = np.full(len(y_test_class), majority_class)

logistic_metrics = pd.DataFrame({
    "Model": ["Majority baseline", "Logistic regression"],
    "Accuracy": [
        accuracy_score(y_test_class, baseline_class_prediction),
        accuracy_score(y_test_class, logistic_prediction),
    ],
    "Precision": [
        precision_score(y_test_class, baseline_class_prediction, zero_division=0),
        precision_score(y_test_class, logistic_prediction, zero_division=0),
    ],
    "Recall": [
        recall_score(y_test_class, baseline_class_prediction, zero_division=0),
        recall_score(y_test_class, logistic_prediction, zero_division=0),
    ],
    "F1": [
        f1_score(y_test_class, baseline_class_prediction, zero_division=0),
        f1_score(y_test_class, logistic_prediction, zero_division=0),
    ],
    "ROC_AUC": [
        np.nan,
        roc_auc_score(y_test_class, logistic_probability),
    ],
})
display(logistic_metrics)
```

```python
logistic_confusion_matrix = confusion_matrix(y_test_class, logistic_prediction)

fig, ax = plt.subplots(figsize=(5.5, 5))
ConfusionMatrixDisplay(
    confusion_matrix=logistic_confusion_matrix,
    display_labels=["pleasant < 50", "pleasant >= 50"],
).plot(ax=ax, values_format="d", colorbar=False)
ax.set_title("Figure 4: Logistic regression - confusion matrix")
plt.tight_layout()
plt.show()

false_positive_rate, true_positive_rate, _ = roc_curve(
    y_test_class, logistic_probability
)
auc_value = roc_auc_score(y_test_class, logistic_probability)

plt.figure(figsize=(6, 5))
plt.plot(false_positive_rate, true_positive_rate, label=f"AUC = {auc_value:.3f}")
plt.plot([0, 1], [0, 1], linestyle="--", label="Random classifier")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("Figure 5: Logistic regression - ROC curve")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
logistic_coefficients = pd.DataFrame({
    "feature": logistic_model.named_steps["preprocessor"].get_feature_names_out(),
    "coefficient": logistic_model.named_steps["model"].coef_[0],
}).sort_values("coefficient", ascending=False)

logistic_numeric_coefficients = logistic_coefficients[
    logistic_coefficients["feature"].str.startswith("numeric__")
].copy()

print("快と判定されやすくする方向（数値・二値特徴量）")
display(logistic_numeric_coefficients.head(10))

print("非快と判定されやすくする方向（数値・二値特徴量）")
display(logistic_numeric_coefficients.tail(10).sort_values("coefficient"))
```

# 5．主成分分析（PCA）

`pleasant`、`intensive`、`familiar`、`warm`、`cold`、`irritating`、`edible`、`disgusting`の8項目を標準化し、PCAを適用する。

寄与率と主成分負荷量を算出し、図6から図8に示す。


```python
PCA_FEATURES = [
    "pleasant", "intensive", "familiar", "warm",
    "cold", "irritating", "edible", "disgusting",
]

pca_data = included[PCA_FEATURES].dropna()
pca_standardized = StandardScaler().fit_transform(pca_data)
pca = PCA(n_components=len(PCA_FEATURES))
pca.fit(pca_standardized)

component_names = [f"PC{i}" for i in range(1, len(PCA_FEATURES) + 1)]
pca_explained_variance = pd.DataFrame({
    "component": component_names,
    "explained_variance_ratio": pca.explained_variance_ratio_,
    "cumulative_ratio": np.cumsum(pca.explained_variance_ratio_),
})
pca_loadings = pd.DataFrame(
    pca.components_.T,
    index=PCA_FEATURES,
    columns=component_names,
)

display(pca_explained_variance)
display(pca_loadings[["PC1", "PC2"]])
print("第1・第2主成分の累積寄与率:", pca.explained_variance_ratio_[:2].sum())
```

# 8．アブレーション分析

完全モデルの性能がどの入力情報から生じるかを確認するため、同一の訓練・テスト分割で説明変数を段階的に追加する。

- A：年齢・性別・臭気物質コード
- B：A＋16種類の質的ラベル
- C：B＋`disgusting`を除く連続知覚評価
- D：C＋`disgusting`（完全モデル）
- E：完全モデルから臭気物質コードを除外

線形回帰の比較を図12、ロジスティック回帰の比較を図13に示す。

```python
ABLATION_CONFIGS = {
    "A": {
        "description": "属性＋臭気物質",
        "numeric": ["age"],
        "categorical": ["molcode", "sex"],
    },
    "B": {
        "description": "A＋質的ラベル",
        "numeric": ["age", *BINARY_ODOR_COLUMNS],
        "categorical": ["molcode", "sex"],
    },
    "C": {
        "description": "B＋連続知覚（disgusting除外）",
        "numeric": [
            "age", *BINARY_ODOR_COLUMNS,
            "intensive", "familiar", "warm", "cold",
            "irritating", "edible",
        ],
        "categorical": ["molcode", "sex"],
    },
    "D": {
        "description": "完全モデル",
        "numeric": [
            "age", *BINARY_ODOR_COLUMNS,
            "intensive", "familiar", "warm", "cold",
            "irritating", "edible", "disgusting",
        ],
        "categorical": ["molcode", "sex"],
    },
    "E": {
        "description": "完全モデル（molcode除外）",
        "numeric": [
            "age", *BINARY_ODOR_COLUMNS,
            "intensive", "familiar", "warm", "cold",
            "irritating", "edible", "disgusting",
        ],
        "categorical": ["sex"],
    },
}


def make_ablation_preprocessor(numeric_features, categorical_features):
    transformers = []

    if numeric_features:
        transformers.append((
            "numeric",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler()),
            ]),
            numeric_features,
        ))

    if categorical_features:
        transformers.append((
            "categorical",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(handle_unknown="ignore")),
            ]),
            categorical_features,
        ))

    return ColumnTransformer(transformers)


linear_ablation_rows = []
logistic_ablation_rows = []

for model_code, config in ABLATION_CONFIGS.items():
    numeric_features = config["numeric"]
    categorical_features = config["categorical"]
    feature_columns = numeric_features + categorical_features

    X_ablation = model_data[feature_columns]
    X_ablation_train = X_ablation.iloc[train_index]
    X_ablation_test = X_ablation.iloc[test_index]

    linear_ablation_model = Pipeline([
        (
            "preprocessor",
            make_ablation_preprocessor(
                numeric_features,
                categorical_features,
            ),
        ),
        ("model", LinearRegression()),
    ])
    linear_ablation_model.fit(X_ablation_train, y_train)
    linear_ablation_prediction = linear_ablation_model.predict(
        X_ablation_test
    )

    linear_ablation_rows.append({
        "Model": model_code,
        "Description": config["description"],
        "MAE": mean_absolute_error(
            y_test,
            linear_ablation_prediction,
        ),
        "RMSE": np.sqrt(mean_squared_error(
            y_test,
            linear_ablation_prediction,
        )),
        "R2": r2_score(
            y_test,
            linear_ablation_prediction,
        ),
    })

    logistic_ablation_model = Pipeline([
        (
            "preprocessor",
            make_ablation_preprocessor(
                numeric_features,
                categorical_features,
            ),
        ),
        (
            "model",
            LogisticRegression(
                max_iter=3000,
                random_state=RANDOM_STATE,
            ),
        ),
    ])
    logistic_ablation_model.fit(
        X_ablation_train,
        y_train_class,
    )
    logistic_ablation_probability = (
        logistic_ablation_model.predict_proba(
            X_ablation_test
        )[:, 1]
    )
    logistic_ablation_prediction = (
        logistic_ablation_probability >= 0.5
    ).astype(int)

    logistic_ablation_rows.append({
        "Model": model_code,
        "Description": config["description"],
        "Accuracy": accuracy_score(
            y_test_class,
            logistic_ablation_prediction,
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_test_class,
            logistic_ablation_prediction,
        ),
        "F1": f1_score(
            y_test_class,
            logistic_ablation_prediction,
            zero_division=0,
        ),
        "ROC_AUC": roc_auc_score(
            y_test_class,
            logistic_ablation_probability,
        ),
    })

linear_ablation_results = pd.DataFrame(linear_ablation_rows)
logistic_ablation_results = pd.DataFrame(logistic_ablation_rows)

display(linear_ablation_results.round(3))
display(logistic_ablation_results.round(3))

plt.figure(figsize=(8, 5))
plt.bar(
    linear_ablation_results["Model"],
    linear_ablation_results["R2"],
)
plt.xlabel("Feature set")
plt.ylabel("R-squared")
plt.title("Figure 12: Linear regression ablation")
plt.ylim(0, 0.7)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(
    logistic_ablation_results["Model"],
    logistic_ablation_results["ROC_AUC"],
)
plt.xlabel("Feature set")
plt.ylabel("ROC-AUC")
plt.title("Figure 13: Logistic regression ablation")
plt.ylim(0.5, 1.0)
plt.tight_layout()
plt.show()
```

# 9．二値化閾値の感度分析

快の境界を40点、50点、60点へ変更し、分類性能が特定の閾値だけに依存していないかを確認する。結果を図14に示す。

```python
threshold_rows = []

for threshold in [40, 50, 60]:
    y_train_threshold = (y_train >= threshold).astype(int)
    y_test_threshold = (y_test >= threshold).astype(int)

    threshold_model = Pipeline([
        ("preprocessor", make_preprocessor()),
        (
            "model",
            LogisticRegression(
                max_iter=3000,
                random_state=RANDOM_STATE,
            ),
        ),
    ])
    threshold_model.fit(X_train, y_train_threshold)

    threshold_probability = threshold_model.predict_proba(
        X_test
    )[:, 1]
    threshold_prediction = (
        threshold_probability >= 0.5
    ).astype(int)

    threshold_rows.append({
        "Threshold": threshold,
        "Positive_Rate": y_test_threshold.mean(),
        "Accuracy": accuracy_score(
            y_test_threshold,
            threshold_prediction,
        ),
        "Balanced_Accuracy": balanced_accuracy_score(
            y_test_threshold,
            threshold_prediction,
        ),
        "Precision": precision_score(
            y_test_threshold,
            threshold_prediction,
            zero_division=0,
        ),
        "Recall": recall_score(
            y_test_threshold,
            threshold_prediction,
            zero_division=0,
        ),
        "F1": f1_score(
            y_test_threshold,
            threshold_prediction,
            zero_division=0,
        ),
        "ROC_AUC": roc_auc_score(
            y_test_threshold,
            threshold_probability,
        ),
    })

threshold_sensitivity = pd.DataFrame(threshold_rows)
display(threshold_sensitivity.round(3))

plt.figure(figsize=(8, 5))
plt.plot(
    threshold_sensitivity["Threshold"],
    threshold_sensitivity["Balanced_Accuracy"],
    marker="o",
    label="Balanced accuracy",
)
plt.plot(
    threshold_sensitivity["Threshold"],
    threshold_sensitivity["ROC_AUC"],
    marker="o",
    label="ROC-AUC",
)
plt.xlabel("pleasant threshold")
plt.ylabel("Score")
plt.title("Figure 14: Threshold sensitivity")
plt.xticks([40, 50, 60])
plt.ylim(0.75, 0.95)
plt.legend()
plt.tight_layout()
plt.show()
```

# 10．補足実験：MNIST

MNISTは嗅覚知覚に関する結論を導く分析ではなく、連続値予測とクラス分類で適切な手法が異なることを確認する補足実験である。

計算量を抑えるため、訓練15,000枚、テスト3,000枚を使用し、PCAで784次元から50次元へ圧縮する。線形回帰の出力を四捨五入した場合と、多クラスロジスティック回帰を比較する。

```python
from tensorflow.keras.datasets import mnist

(mnist_X_train_all, mnist_y_train_all), (mnist_X_test_all, mnist_y_test_all) = mnist.load_data()

rng = np.random.default_rng(RANDOM_STATE)
train_sample = rng.choice(len(mnist_X_train_all), 15000, replace=False)
test_sample = rng.choice(len(mnist_X_test_all), 3000, replace=False)

mnist_X_train = mnist_X_train_all[train_sample].reshape(-1, 784).astype("float32") / 255.0
mnist_y_train = mnist_y_train_all[train_sample]
mnist_X_test = mnist_X_test_all[test_sample].reshape(-1, 784).astype("float32") / 255.0
mnist_y_test = mnist_y_test_all[test_sample]

mnist_pca = PCA(n_components=50, svd_solver="randomized", random_state=RANDOM_STATE)
mnist_X_train_pca = mnist_pca.fit_transform(mnist_X_train)
mnist_X_test_pca = mnist_pca.transform(mnist_X_test)

print("MNIST訓練データ:", mnist_X_train.shape)
print("MNISTテストデータ:", mnist_X_test.shape)
print("50主成分の累積寄与率:", mnist_pca.explained_variance_ratio_.sum())
```

```python
mnist_linear_model = LinearRegression()
mnist_linear_model.fit(mnist_X_train_pca, mnist_y_train)

mnist_linear_raw_prediction = mnist_linear_model.predict(mnist_X_test_pca)
mnist_linear_class_prediction = np.clip(
    np.rint(mnist_linear_raw_prediction), 0, 9
).astype(int)

mnist_linear_mae = mean_absolute_error(
    mnist_y_test, mnist_linear_raw_prediction
)
mnist_linear_accuracy = accuracy_score(
    mnist_y_test, mnist_linear_class_prediction
)

print("線形回帰のMAE:", mnist_linear_mae)
print("線形回帰を分類器として使った正解率:", mnist_linear_accuracy)
```

```python
mnist_logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=RANDOM_STATE,
)
mnist_logistic_model.fit(mnist_X_train_pca, mnist_y_train)

mnist_logistic_prediction = mnist_logistic_model.predict(mnist_X_test_pca)
mnist_logistic_accuracy = accuracy_score(
    mnist_y_test, mnist_logistic_prediction
)

print("ロジスティック回帰の正解率:", mnist_logistic_accuracy)

fig, ax = plt.subplots(figsize=(9, 8))
ConfusionMatrixDisplay.from_predictions(
    mnist_y_test,
    mnist_logistic_prediction,
    ax=ax,
    colorbar=False,
)
ax.set_title("Figure 15: MNIST logistic regression - confusion matrix")
plt.tight_layout()
plt.show()

mnist_metrics = pd.DataFrame({
    "Model": ["Linear regression + rounding", "Logistic regression"],
    "Accuracy": [mnist_linear_accuracy, mnist_logistic_accuracy],
})
display(mnist_metrics)
```
